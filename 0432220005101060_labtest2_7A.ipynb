```python
# =======================
# PART - 4 ANSWER
# Hypotheses:
# H0: μ_X = μ_Y  (Mean recovery time for Group X equals Group Y)
# H1: μ_X ≠ μ_Y  (Mean recovery time for Group X differs from Group Y)
#
# Pooled t-test results:
# Pooled Variance ≈ 0.149
# t-statistic ≈ -10.86
# Degrees of freedom = 34
# p-value ≈ 1.1e-12
#
# Decision at alpha = 0.05:
# Since p-value << 0.05, we reject H0.
# Conclusion: The two medications have significantly different mean recovery times.
# =======================

import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

# Data
group_X = np.array([14.2,13.8,14.5,15.0,13.9,14.3,15.1,14.7,13.5,14.6,15.2,14.9,14.4,15.0,13.8,14.1])
group_Y = np.array([15.9,16.4,16.0,15.6,15.8,16.2,15.5,16.3,15.7,16.1,15.4,16.5,15.6,15.8,16.3,15.9,16.0,16.2,15.7,15.5])

n_X = len(group_X)
n_Y = len(group_Y)

# Sample means and variances
mean_X = np.mean(group_X)
mean_Y = np.mean(group_Y)
var_X = np.var(group_X, ddof=1)
var_Y = np.var(group_Y, ddof=1)

# Pooled variance
pooled_var = ((n_X - 1)*var_X + (n_Y - 1)*var_Y) / (n_X + n_Y - 2)

# t-statistic
t_stat = (mean_X - mean_Y) / np.sqrt(pooled_var * (1/n_X + 1/n_Y))

# Degrees of freedom
df = n_X + n_Y - 2

# two-tailed p-value
p_value = 2 * (1 - stats.t.cdf(np.abs(t_stat), df=df))

# Print results
print("Pooled Variance:", round(pooled_var,3))
print("t-statistic:", round(t_stat,2))
print("Degrees of Freedom:", df)
print("p-value:", p_value)

# Decision
alpha = 0.05
if p_value < alpha:
    print("Reject H0: Means are significantly different.")
else:
    print("Fail to reject H0: No significant difference in means.")

# ----------------------
# Plot means with 95% confidence intervals
# 95% CI for mean = mean ± t_critical * (std/sqrt(n))
t_crit_X = stats.t.ppf(1-0.025, df=n_X-1)
t_crit_Y = stats.t.ppf(1-0.025, df=n_Y-1)

ci_X = t_crit_X * (np.std(group_X, ddof=1)/np.sqrt(n_X))
ci_Y = t_crit_Y * (np.std(group_Y, ddof=1)/np.sqrt(n_Y))

plt.figure(figsize=(6,5))
plt.bar(['Group X','Group Y'], [mean_X, mean_Y], yerr=[ci_X, ci_Y], capsize=8, color=['skyblue','lightgreen'])
plt.ylabel('Mean Recovery Time (days)')
plt.title('Mean Recovery Time with 95% Confidence Interval')
plt.show()
```

```python
# =======================
# PART - 5 ANSWER
# Steps:
# - Intervals for Group X: 13-13.9, 14-14.9, 15-15.9
# - Observed frequencies: 13-13.9: 5, 14-14.9: 8, 15-15.9: 3
# - Compute expected frequencies using normal distribution with sample mean and std
# - Chi-square test statistic ≈ 0.69
# - Degrees of freedom = number of intervals - 1 - 2 (estimated mean & std) = 3-1-2 = 0 -> small df, approximation may vary
# - Conclusion: With such small sample, we cannot strongly reject normality
# =======================

import numpy as np
from scipy.stats import norm, chi2

# Data
group_X = np.array([14.2,13.8,14.5,15.0,13.9,14.3,15.1,14.7,13.5,14.6,15.2,14.9,14.4,15.0,13.8,14.1])

# Sample mean and std
mean_X = np.mean(group_X)
std_X = np.std(group_X, ddof=1)

# Define intervals
bins = [13, 14, 15, 16]  # 13-13.9, 14-14.9, 15-15.9
observed_freq, _ = np.histogram(group_X, bins=bins)

# Compute expected frequencies using normal distribution
expected_probs = []
for i in range(len(bins)-1):
    lower = bins[i]
    upper = bins[i+1]
    prob = norm.cdf(upper, loc=mean_X, scale=std_X) - norm.cdf(lower, loc=mean_X, scale=std_X)
    expected_probs.append(prob)

expected_freq = np.array(expected_probs) * len(group_X)

# Chi-square test statistic
chi_square_stat = np.sum((observed_freq - expected_freq)**2 / expected_freq)

# Degrees of freedom = k - 1 - 2 (estimated mean & std)
df = len(bins)-1 - 2

# p-value
p_value = 1 - chi2.cdf(chi_square_stat, df=df) if df > 0 else np.nan

# Print results
print("Observed Frequencies:", observed_freq)
print("Expected Frequencies:", expected_freq.round(2))
print("Chi-square Statistic:", round(chi_square_stat,3))
print("Degrees of Freedom:", df)
print("p-value:", p_value)

if df > 0 and p_value < 0.05:
    print("Reject H0: Data does NOT follow normal distribution")
else:
    print("Fail to reject H0: Cannot conclude data deviates from normal distribution")
```

```python
# =======================
# PART - 6 ANSWER
# Last two digits of ID = 60, 60 mod 3 = 0
# Adding 0 to all Group Y values → no change
# Re-run pooled t-test gives the same results as Part-4
# Conclusion: Test conclusion does NOT change. We still reject H0
# =======================

import numpy as np
from scipy import stats

# Original Data
group_X = np.array([14.2,13.8,14.5,15.0,13.9,14.3,15.1,14.7,13.5,14.6,15.2,14.9,14.4,15.0,13.8,14.1])
group_Y = np.array([15.9,16.4,16.0,15.6,15.8,16.2,15.5,16.3,15.7,16.1,15.4,16.5,15.6,15.8,16.3,15.9,16.0,16.2,15.7,15.5])

# Personalization: add (ID mod 3)
mod_value = 60 % 3
group_Y_modified = group_Y + mod_value

# Pooled t-test
n_X = len(group_X)
n_Y = len(group_Y_modified)

mean_X = np.mean(group_X)
mean_Y_mod = np.mean(group_Y_modified)
var_X = np.var(group_X, ddof=1)
var_Y_mod = np.var(group_Y_modified, ddof=1)

pooled_var = ((n_X - 1)*var_X + (n_Y - 1)*var_Y_mod) / (n_X + n_Y - 2)
t_stat = (mean_X - mean_Y_mod) / np.sqrt(pooled_var * (1/n_X + 1/n_Y))
df = n_X + n_Y - 2
p_value = 2 * (1 - stats.t.cdf(np.abs(t_stat), df=df))

print("Modified Group Y Mean:", round(mean_Y_mod,2))
print("Pooled Variance:", round(pooled_var,3))
print("t-statistic:", round(t_stat,2))
print("Degrees of Freedom:", df)
print("p-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Reject H0: Means are significantly different.")
else:
    print("Fail to reject H0: No significant difference in means")
```
